## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
import pandas as pd
import json
# from io import StringIO is removed as we are reading files directly

# 1. File Path Definitions
POSTS_FILE = '/content/final_posts.json'
SPLIT_FILE = '/content/data_identification.csv'
EMOTION_FILE = '/content/emotion.csv'

# 2. Processing final_posts.json

def flatten_post_data(data):
    """Extracts post_id and text from the nested JSON structure."""
    posts = []
    for item in data:
        # Safely access the deeply nested 'post_id' and 'text' fields
        post_data = item.get("root", {}).get("_source", {}).get("post", {})
        if post_data:
            posts.append({
                'post_id': post_data.get('post_id'),
                'text': post_data.get('text'),
                # You might also want to keep 'hashtags' for feature engineering later
                'hashtags': post_data.get('hashtags', [])
            })
    return pd.DataFrame(posts)

# Load the posts data
print(f"Loading posts from {POSTS_FILE}")
with open(POSTS_FILE, 'r') as f:
    raw_json_data = json.load(f)

df_posts = flatten_post_data(raw_json_data)

print("--- DataFrame from final_posts.json (df_posts) ---")
print(df_posts.head())
print(f"Total posts loaded: {len(df_posts)}")
print("\n")

# 3. Loading CSV Files

# Load split identification data
print(f"Loading split identification from {SPLIT_FILE}")
df_split = pd.read_csv(SPLIT_FILE)
df_split.rename(columns={'id': 'post_id'}, inplace=True)

print("--- DataFrame from data_identification.csv (df_split) ---")
print(df_split.head())
print("\n")

# Load emotion labels data
print(f"Loading emotion labels from {EMOTION_FILE}")
df_emotion = pd.read_csv(EMOTION_FILE)
df_emotion.rename(columns={'id': 'post_id'}, inplace=True)

print("--- DataFrame from emotion.csv (df_emotion) ---")
print(df_emotion.head())
print("\n")

# 4. Merging DataFrames

# Merge 1: Posts with split identification
df_merged = pd.merge(df_posts, df_split, on='post_id', how='left')

# Merge 2: Add emotion labels (left join ensures all posts are kept, even those without labels)
# The posts in the 'test' split will have NaN in the 'emotion' column.
final_df = pd.merge(df_merged, df_emotion, on='post_id', how='left')

print("--- Final Merged DataFrame ---")
print(final_df.head()) # Use .head() as the full dataset might be large
print(f"Total merged records: {len(final_df)}")
print("\n")

# 5. Splitting into Train and Test Sets

# Separate the dataset based on the 'split' column
df_train = final_df[final_df['split'] == 'train'].reset_index(drop=True)
df_test = final_df[final_df['split'] == 'test'].reset_index(drop=True)

# Drop the 'split' column from the final sets as it's no longer needed for modeling
df_train = df_train.drop(columns=['split'])
df_test = df_test.drop(columns=['split'])

print("--- Training Data (df_train) ---")
print(df_train.head())
print(f"\nTraining set size: {len(df_train)} posts")

print("\n--- Testing Data (df_test) ---")
print(df_test.head())
print(f"\nTesting set size: {len(df_test)} posts")

# Check for the expected missing labels in the test set
print(f"\nNaN count in test set emotion column: {df_test['emotion'].isna().sum()}")
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
import nltk
nltk.download('stopwords')
nltk.download('wordnet')

stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

def preprocess_text(text):
    """Clean and normalize text using lowercasing, regex, and lemmatization."""
    if not isinstance(text, str):
        return ""

    # 1. Lowercase
    text = text.lower()

    # 2. Remove punctuation, special characters, and numbers
    text = re.sub(r'[^a-z\s]', '', text)

    # 3. Tokenization and Lemmatization
    tokens = text.split()

    # 4. Remove Stop Words and Lemmatize
    tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]

    return " ".join(tokens)

# Apply preprocessing to both datasets
df_train['clean_text'] = df_train['text'].apply(preprocess_text)
df_test['clean_text'] = df_test['text'].apply(preprocess_text)

print("--- Preprocessed Training Text Example ---")
print(df_train[['text', 'clean_text']].head())
print("\n")

# --- 3. TF-IDF Vectorization ---

# Initialize the TF-IDF Vectorizer
# - max_features: Limits the number of features (vocabulary size)
# - ngram_range: Uses both single words (1-gram) and pairs of words (2-gram)
tfidf_vectorizer = TfidfVectorizer(max_features=5000, ngram_range=(1, 2))

# 3.1. Fit and Transform the Training Data
# The vectorizer learns the vocabulary ONLY from the training data.
X_train_tfidf = tfidf_vectorizer.fit_transform(df_train['clean_text'])
y_train = df_train['emotion'] # Target variable

# 3.2. Transform the Test Data
# Use the SAME vectorizer (do NOT call fit_transform) to avoid data leakage.
X_test_tfidf = tfidf_vectorizer.transform(df_test['clean_text'])

print(f"TF-IDF Matrix created.")
print(f"Training Features Matrix Shape (Posts x Features): {X_train_tfidf.shape}")
print(f"Testing Features Matrix Shape (Posts x Features): {X_test_tfidf.shape}")

# X_train_tfidf is now your feature matrix ready for model training.
# X_test_tfidf is your feature matrix ready for prediction.

# --- 4. Next Step: Model Training ---
# Example: Using a simple Logistic Regression classifier

from sklearn.linear_model import LogisticRegression

# Since emotion labels are strings, we need to convert them to numerical format
y_train_encoded = y_train.astype('category').cat.codes

model = LogisticRegression(max_iter=1000)
model.fit(X_train_tfidf, y_train_encoded)

# The model is now trained and ready to predict on X_test_tfidf.
# For a small dataset like this, training metrics will not be meaningful,
# but for your full dataset, this is the blueprint!
print("\nModel trained successfully using Logistic Regression.")
```

## 3. Model Implementation Steps

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
### Add the code related to the model implementation steps in cells inside this section
import torch
import numpy as np
import pandas as pd
from datasets import Dataset
from transformers import AutoTokenizer, AutoModelForSequenceClassification, TrainingArguments, Trainer
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score
from sklearn.preprocessing import LabelEncoder

# 1. Label Encoding and Data Conversion

# 1.1 Encode Text Labels to Integers (required for the model)
# We must fit the encoder ONLY on the training labels to avoid data leakage.
label_encoder = LabelEncoder()
df_train['labels'] = label_encoder.fit_transform(df_train['emotion'])

# Store the mapping for decoding predictions later
label_map = dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_)))
num_labels = len(label_map)
print(f"Emotion Label Mapping: {label_map}")

# 1.2 Convert Pandas DataFrames to Hugging Face Dataset objects
train_dataset = Dataset.from_pandas(df_train[['text', 'labels', 'post_id']])
# Test set needs to maintain post_id for submission
test_dataset = Dataset.from_pandas(df_test[['text', 'post_id']])

# 2. Tokenizer and Model Loading

# Use a pre-trained model
MODEL_NAME = "distilbert-base-uncased"

print(f"Loading Tokenizer: {MODEL_NAME}")
# The tokenizer converts raw text into numerical IDs the model understands.
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

print(f"Loading Model: {MODEL_NAME} with {num_labels} classification heads.")
# AutoModelForSequenceClassification loads the pre-trained weights
# and adds a classification layer specific to our number of emotion classes.
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME, num_labels=num_labels)

# 3. Tokenization Function

def tokenize_function(examples):
    """Applies the tokenizer to the text column of the dataset."""
    # Truncation=True ensures sequences are cut to the max length the model supports (512 for BERT)
    return tokenizer(examples["text"], truncation=True, padding="max_length")

# Apply tokenization to the datasets
tokenized_train_dataset = train_dataset.map(tokenize_function, batched=True)
tokenized_test_dataset = test_dataset.map(tokenize_function, batched=True)

# Remove original text column since it's now embedded in tokenized data
tokenized_train_dataset = tokenized_train_dataset.remove_columns(["text", "post_id"])
tokenized_test_dataset = tokenized_test_dataset.remove_columns(["text", "post_id"])

# Rename 'labels' column to 'label' for the Hugging Face Trainer
tokenized_train_dataset = tokenized_train_dataset.rename_column("labels", "label")
tokenized_train_dataset.set_format("torch")

# 4. Training Setup and Metrics

# Define a function to compute metrics during training (validation)
def compute_metrics(eval_pred):
    """Computes standard classification metrics."""
    logits, labels = eval_pred
    predictions = np.argmax(logits, axis=-1)

    # Competition scoring often uses F1-score or Accuracy
    accuracy = accuracy_score(labels, predictions)
    f1 = f1_score(labels, predictions, average='weighted')

    return {"accuracy": accuracy, "f1_weighted": f1}

# Define Training Arguments
# These are the hyperparameters for the fine-tuning process.
training_args = TrainingArguments(
    output_dir="./results",                     # Output directory for checkpoints
    num_train_epochs=3,                        # Number of training epochs (try 3-5)
    per_device_train_batch_size=4,            # Batch size per GPU/CPU for training
    per_device_eval_batch_size=4,             # Batch size for evaluation
    warmup_steps=500,                          # Number of warmup steps for learning rate scheduler
    weight_decay=0.01,                         # Strength of weight decay
    logging_dir='./logs',                      # Directory for logs
    eval_strategy="epoch",                     # Evaluate every 'eval_steps'
    save_strategy="epoch",                     # Save checkpoint every epoch
    load_best_model_at_end=True,               # Load the best model found during training
    report_to="none"                           # Disable integration with services like WandB
)

# Initialize the Trainer
trainer = Trainer(
    model=model,                              # The instantiated model to be trained
    args=training_args,                       # The training arguments
    train_dataset=tokenized_train_dataset,    # Training dataset
    eval_dataset=tokenized_train_dataset,     # Using train as eval for simplicity (in reality, use a validation split)
    tokenizer=tokenizer,                      # The tokenizer
    compute_metrics=compute_metrics           # The function to compute metrics
)

# --- 6. Model Training ---
print("\n" + "="*50)
print("STARTING BERT MODEL FINE-TUNING")
print("="*50)
trainer.train()
print("Training Complete.")

# --- 7. Prediction on Test Set ---

# Make predictions on the competition test set
print("\nMaking predictions on the test set...")
predictions = trainer.predict(tokenized_test_dataset)
logits = predictions.predictions

# Get the predicted class index (highest logit value)
predicted_label_indices = np.argmax(logits, axis=1)

# Decode indices back to original emotion names
id_to_label = {v: k for k, v in label_map.items()}
predicted_emotions = [id_to_label[i] for i in predicted_label_indices]

# --- 8. Create Submission File (sampleSubmission.csv format) ---

submission_df = pd.DataFrame({
    'id': df_test['post_id'],
    'emotion': predicted_emotions
})

SUBMISSION_FILE = '/content/bert_submission.csv'
submission_df.to_csv(SUBMISSION_FILE, index=False)

print("\n" + "="*50)
print(f"Prediction Complete. Submission file saved to {SUBMISSION_FILE}")
print("--- Sample Submission Output ---")
print(submission_df.head())
print("="*50)
```
